1. Temperature Variables (T2M, MAX, MIN)
T2M & T2M_MIN (0.95): There is a very high positive correlation here. This suggests that in Tanzania, the average daily temperature is driven heavily by how cool it gets at night.

T2M_RANGE & RH2M (-0.79): A strong negative correlation.

The Tanzania Context: When humidity (RH2M) is high (Rainy Season), the temperature range is small. When it is dry (Dry Season), you get much hotter days and cooler nights, causing the "range" to increase.

2. Precipitation (PRECTOTCORR)
Correlation with Humidity (0.51): As expected, as rainfall increases, relative humidity also increases.

Correlation with Temp Range (-0.52): When it rains, the gap between the day's high and low temperatures shrinks. This is because cloud cover during the rainy season traps heat at night and blocks intense sun during the day.

3. Humidity (RH2M & QV2M)
RH2M (Relative Humidity) vs. QV2M (Specific Humidity) (0.71): These usually move together. In Tanzania’s coastal or lake regions, both will be high.

Specific Humidity (QV2M) & T2M_MIN (0.89): This is a very strong link. It shows thaat "wet" air is excellent at holding onto heat at night, preventing temperatures from dropping too low.

4. Surface Pressure (PS)
PS vs. T2M (-0.84): This is a very strong negative correlation. In meteorology, as air heats up, it expands and becomes less dense, leading to lower surface pressure.

PS vs. T2M_MIN (-0.84): Similarly, higher pressure is strongly associated with cooler nights.

5. Wind Speed (WS2M & WS2M_MAX)
WS2M vs. WS2M_MAX (0.95): These are nearly identical. If the average wind speed is high, the gusts (max speed) are also high.

Wind & Pressure (0.57): There is a moderate positive link here. High-pressure systems often bring sustained winds to the region.

6. Time Factors (DOY, Month, Year)
DOY & Month (1.0): This is a perfect correlation because the "Day of Year" and "Month" are just two different ways of measuring the same timeline.

Year (near 0): The correlation between "Year" and variables like Temperature or Rain is very low (e.g., 0.059 for T2M). This suggests that over the specific period this data was collected, there hasn't been a massive, steady upward or downward trend—the weather is mostly following seasonal cycles rather than a straight line change.
