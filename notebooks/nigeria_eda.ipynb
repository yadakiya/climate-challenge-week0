```python
df["Date"] = pd.to_datetime(df["YEAR"] * 1000 + df["DOY"], format="%Y%j")
```

```python
df["Month"] = df["Date"].dt.month
```

```python
df.replace(-999, np.nan, inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
z_scores = np.abs(zscore(df.select_dtypes(include=[np.number]).dropna()))
```

```python
df.ffill(inplace=True)
```

```python
df.groupby("Month")["T2M"].mean().plot()
plt.title("Average Monthly Temperature")
plt.show()
```

```python
df.to_csv("../data/nigeria_clean.csv", index=False)
```

```python
df.groupby("Month")["PRECTOTCORR"].sum().plot(kind="bar")
plt.title("Monthly Rainfall")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()
```

Overview

This heatmap shows the correlation between different climate variables in the Nigeria dataset.
Correlation values range from -1 to +1:

+1 → strong positive relationship
-1 → strong negative relationship
0 → no relationship
🌡 Temperature Relationships
T2M, T2M_MAX, and T2M_MIN show strong positive correlation (~0.85)
👉 This means:
When average temperature increases, maximum and minimum temperatures also increase
These variables are closely related and consistent
T2M_RANGE has:
negative correlation with T2M_MIN (-0.57)
👉 When minimum temperature increases, the temperature range decreases
🌧 Precipitation Relationships
PRECTOTCORR (rainfall) has:
Weak to moderate correlation with temperature variables
👉 Rainfall is relatively independent of temperature changes
Moderate positive correlation with:
RH2M (humidity) (~0.35)
👉 More rainfall is associated with higher humidity
💧 Humidity (RH2M)
Strong negative correlation with T2M_RANGE (-0.81)
👉 When humidity increases, temperature variation decreases
Moderate positive correlation with:
QV2M (~0.67) (specific humidity)
👉 Indicates consistency between humidity-related variables
💨 Wind Speed (WS2M, WS2M_MAX)
WS2M and WS2M_MAX have very strong correlation (~0.93)
👉 Wind speed and maximum wind speed behave similarly
Weak correlation with temperature
👉 Wind is mostly independent of temperature changes
🧠 Pressure (PS)
Negative correlation with temperature (-0.68)
👉 Higher temperatures are associated with lower pressure
Slight positive relationship with wind speed
📅 Time Variables (YEAR, MONTH, DOY)
Weak correlations with most variables
👉 Climate variables do not strongly depend on simple time indices
🔥 Key Insights
Temperature variables are highly interdependent
Humidity strongly affects temperature variation
Rainfall is more connected to humidity than temperature
Wind speed variables are internally consistent but independent of temperature
Pressure decreases as temperature increases
📌 Conclusion

The correlation analysis reveals that:

Temperature-related variables move together strongly
Humidity plays a key role in stabilizing temperature
Rainfall is more influenced by atmospheric moisture than temperature
Some variables (like wind) behave independently

This helps in understanding climate patterns and dependencies in Nigeria, which is useful for further modeling and prediction
