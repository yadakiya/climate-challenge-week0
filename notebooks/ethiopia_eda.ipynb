```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore
```

🌡️ 1. Temperature Relationships
T2M and T2M_MAX (0.65) → strong positive correlation
👉 When average temperature increases, max temperature also increases.
T2M and T2M_MIN (0.72) → strong positive correlation
👉 Warm days usually have higher minimum temperatures too.
T2M_RANGE and T2M_MIN (-0.69) → strong negative correlation
👉 Higher minimum temperature reduces daily temperature variation.
💧 2. Humidity & Rainfall
RH2M and QV2M (0.9) → very strong positive correlation
👉 Humidity variables move together strongly.
T2M_RANGE and QV2M (-0.89) → strong negative correlation
👉 High humidity reduces temperature variation.
PRECTOTCORR and RH2M (0.5) → moderate positive correlation
👉 More rain usually means higher humidity.
🌬️ 3. Wind Patterns
WS2M and WS2M_MAX (0.94) → very strong positive correlation
👉 Maximum wind speed strongly depends on average wind speed.
Wind vs Humidity (negative correlations)
👉 Strong wind often reduces humidity slightly.
🧭 4. Time Variables
DOY (Day of Year) has negative correlation with temperature variables
👉 Suggests seasonal climate variation.
Month shows similar behavior to DOY
👉 Confirms seasonal patterns in Ethiopia’s climate.
📉 5. Pressure (PS)
Weak correlation with most variables
👉 Surface pressure is relatively independent in this dataset.
📌 Key Conclusions

✔ Temperature variables are strongly connected
✔ Humidity variables are tightly linked
✔ Wind speed is very consistent across measurements
✔ Rainfall increases humidity but reduces temperature variation
✔ Strong seasonal patterns exist (DOY and Month effects)
